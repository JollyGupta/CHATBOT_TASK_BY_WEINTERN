```python
import gradio as gr
import re

def chatbot(message, history):
    message = message.lower()

    # Greeting
    if re.search(r"\b(hi|hello|hey)\b", message):
        return "Hello Welcome to our store How may i help you!"

    # Category
    elif re.search(r"\b(category|products|show)\b", message):
        return "We sell Laptop and Mobile. Please choose one."

    # Laptop brands
    elif re.search(r"\blaptop\b", message):
        return "Available laptop brands: HP, Dell"

    # Mobile brands
    elif re.search(r"\bmobile\b", message):
        return "Available mobile brands: Samsung, iPhone"

    # Prices
    elif re.search(r"\bhp\b", message):
        return "HP Laptop price is ₹50,000"

    elif re.search(r"\bdell\b", message):
        return "Dell Laptop price is ₹55,000"

    elif re.search(r"\bsamsung\b", message):
        return "Samsung Mobile price is ₹30,000"

    elif re.search(r"\biphone\b", message):
        return "iPhone price is ₹70,000"

    # Exit
    elif re.search(r"\b(bye|exit|thanks)\b", message):
        return "Thank you for visiting! 👋"

    # Fallback
    else:
        return "Sorry, I didn't understand. Please choose a product."

gr.ChatInterface(chatbot).launch()
```

```python
'''In real NLP systems:

Known intents → trained / rule-based

Unknown input → fallback / default handler

Fallback is always outside training data.'''
```

```python
import gradio as gr
import re

# -------- Intent Keywords --------
INTENTS = {
    "greeting": [r"\bhi\b", r"\bhello\b", r"\bhey\b"],
    "category": [r"\bcategory\b", r"\bproducts\b", r"\bshow\b"],
    "laptop": [r"\blaptop\b"],
    "mobile": [r"\bmobile\b"],
    "price_hp": [r"\bhp\b"],
    "price_dell": [r"\bdell\b"],
    "price_samsung": [r"\bsamsung\b"],
    "price_iphone": [r"\biphone\b"],
    "exit": [r"\bbye\b", r"\bexit\b", r"\bthanks\b"]
}

# -------- Intent Detection --------
def detect_intent(message):
    for intent, patterns in INTENTS.items():
        for pattern in patterns:
            if re.search(pattern, message):
                return intent
    return "fallback"

# -------- Chatbot Logic --------
def chatbot(message, history):
    message = message.lower()
    intent = detect_intent(message)

    if intent == "greeting":
        return "Hello  Welcome to our store! How may I help you?"

    elif intent == "category":
        return "We sell **Laptop** and **Mobile**. Please choose one."

    elif intent == "laptop":
        return "Available laptop brands: **HP, Dell**"

    elif intent == "mobile":
        return "Available mobile brands: **Samsung, iPhone**"

    elif intent == "price_hp":
        return "HP Laptop price is ₹50,000"

    elif intent == "price_dell":
        return "Dell Laptop price is ₹55,000"

    elif intent == "price_samsung":
        return "Samsung Mobile price is ₹30,000"

    elif intent == "price_iphone":
        return "iPhone price is ₹70,000"

    elif intent == "exit":
        return "Thank you for visiting!Have a great day."

    else:
        return (
            "Sorry, I didn’t understand that\n"
            "You can ask about:\n"
            "• Products\n"
            "• Laptop or Mobile\n"
            "• Prices"
        )

# -------- Launch UI --------
gr.ChatInterface(chatbot, title="Smart Store Chatbot").launch()
```

```python
import gradio as gr
import re

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

# ---------------- TRAINING DATA (INTENTS) ----------------
training_sentences = [
    "hi", "hello", "hey",
    "show products", "what do you sell", "category",
    "laptop", "show laptop",
    "mobile", "show mobile",
    "hp price", "price of hp laptop",
    "dell price", "price of dell laptop",
    "samsung price",
    "iphone price",
    "bye", "exit", "thanks"
]

training_labels = [
    "greeting", "greeting", "greeting",
    "category", "category", "category",
    "laptop", "laptop",
    "mobile", "mobile",
    "hp_price", "hp_price",
    "dell_price", "dell_price",
    "samsung_price",
    "iphone_price",
    "exit", "exit", "exit"    #Because multiple input sentences can belong to the same intent,
]


training_sentences += [
    "samsung",
    "iphone",
    "hp",
    "dell"
]

training_labels += [
    "samsung_price",
    "iphone_price",
    "hp_price",
    "dell_price"
]

# ---------------- ML MODEL ----------------

'''Converts text into numeric features
ML models cannot read text, only numbers'''

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(training_sentences)

#Intent Learning (Logistic Regression)
model = LogisticRegression()
model.fit(X, training_labels)


# ---------------- CONTEXT ----------------
#context = {}

# ---------------- CHATBOT LOGIC ----------------
def chatbot(message, history):
    message = message.lower()

    X_test = vectorizer.transform([message])
    intent = model.predict(X_test)[0]

    '''User types anything → Model predicts one intent label
       This replaces regex logic completely.
       Intent names are developer-defined labels, not user input.
       '''

    if intent == "greeting":
        return "Hello  Welcome to our store! How may I help you?"

    elif intent == "category":
        return "We sell **Laptop** and **Mobile**."

    elif intent == "laptop":
        #context["product"] = "laptop"
        return "Available laptop brands: **HP, Dell**"

    elif intent == "mobile":
        #context["product"] = "mobile"
        return "Available mobile brands: **Samsung, iPhone**"

    elif intent == "hp_price":
        return " HP Laptop price is ₹50,000"

    elif intent == "dell_price":
        return "Dell Laptop price is ₹55,000"

    elif intent == "samsung_price":
        return "Samsung Mobile price is ₹30,000"

    elif intent == "iphone_price":
        return "iPhone price is ₹70,000"

    elif intent == "exit":
        return "Thank you for visiting!Have a great day."

    else:
        return (
            "Sorry, I didn’t understand that \n"
            "You can ask about products, laptops, mobiles, or prices."
        )

# ---------------- LAUNCH UI ----------------
gr.ChatInterface(
    chatbot,
    title="ML-Based Rule Chatbot (Gradio)"
).launch()
```

```python
import gradio as gr
import spacy

# Load spaCy model
nlp = spacy.load("en_core_web_sm")

# -------- PRICE DATA --------
PRICES = {
    "laptop": {
        "hp": 50000,
        "dell": 55000
    },
    "mobile": {
        "samsung": 30000,
        "iphone": 70000
    }
}

# -------- CONTEXT (MEMORY) --------
context = {
    "product": None,
    "brand": None
}

# -------- CHATBOT LOGIC --------
def chatbot(message, history):
    message = message.lower()
    doc = nlp(message)

    # -------- EXIT --------
    if any(word in message for word in ["bye", "exit", "thanks"]):
        context["product"] = None
        context["brand"] = None
        return "Thank you for visiting!Have a great day."

    # -------- GREETING --------
    if any(word in message for word in ["hi", "hello", "hey"]):
        return "Hello Welcome to our store! How may I help you?"

    # -------- CATEGORY --------
    if "product" in message or "sell" in message or "category" in message:
        return "We sell **Laptop** and **Mobile**."

    # -------- PRODUCT DETECTION --------
    if "laptop" in message:
        context["product"] = "laptop"
        context["brand"] = None
        return "Available laptop brands: **HP, Dell**"

    if "mobile" in message:
        context["product"] = "mobile"
        context["brand"] = None
        return "Available mobile brands: **Samsung, iPhone**"

    # -------- BRAND DETECTION --------
    for brand in ["hp", "dell", "samsung", "iphone"]:
        if brand in message:
            context["brand"] = brand
            return f"{brand.capitalize()} selected Ask me the price."

    # -------- PRICE --------
    if "price" in message or "cost" in message:
        product = context.get("product")
        brand = context.get("brand")

        if not product:
            return "Please select a product first (Laptop or Mobile)."

        if not brand:
            return "Please select a brand first."

        price = PRICES[product][brand]
        emoji = "💻" if product == "laptop" else "📱"
        return f"{emoji} {brand.capitalize()} {product} price is ₹{price}"

    # -------- FALLBACK --------
    return (
        "Sorry 😔 I didn’t understand.\n"
        "You can ask about products, brands, or prices."
    )

# -------- LAUNCH UI --------
gr.ChatInterface(
    chatbot,
    title="Context-Based NLP Chatbot (spaCy + Gradio)"
).launch()
```
